## Q1. Sensor Readings and Vectorization

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2

print("Original temperature:", temperature)
print("Corrected temperature:", corrected_temperature)
```

```python
fahrenheit = (9 / 5) * corrected_temperature + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

```python
above_32 = corrected_temperature[corrected_temperature > 32]

print("Readings greater than 32°C:", above_32)
print("Number of readings greater than 32°C:", len(above_32))
```

### Explanation

Vectorization allows NumPy to perform operations on the whole array at once instead of using a loop for every value. Boolean indexing allows us to directly select the values that satisfy a condition. These operations are shorter and generally faster than explicitly iterating through every element.

## Q2. Daily Steps Analysis

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps array:")
print(steps)
```

```python
print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
```

```python
day_totals = np.sum(steps, axis=0)

print("Total steps for each day:", day_totals)
```

```python
user_totals = np.sum(steps, axis=1)

print("Total steps for each user:", user_totals)
```

```python
max_index = np.argmax(steps)
max_position = np.unravel_index(max_index, steps.shape)

print("Position of maximum value:", max_position)
print("Maximum value:", steps[max_position])
```

## Q3. NumPy Slicing, Views, Copies and Reshaping

```python
original = np.array([1, 2, 3, 4, 5, 6])

print("Original:", original)
```

```python
subset = original[1:5]

subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)
```

```python
copied_array = original[1:5].copy()

copied_array[0] = 500

print("Original after modifying copied array:", original)
print("Copied array:", copied_array)
```

```python
matrix = np.arange(1, 13).reshape(3, 4)

print("3 x 4 matrix:")
print(matrix)
```

```python
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])
```

```python
flattened = matrix.flatten()
raveled = matrix.ravel()

print("Using flatten():", flattened)
print("Using ravel():", raveled)
```

```python
raveled = matrix.ravel()

raveled[0] = 100

print("Raveled array:", raveled)
print("Original matrix after modifying ravel():")
print(matrix)
```

```python
matrix = np.arange(1, 13).reshape(3, 4)

flattened = matrix.flatten()

flattened[0] = 200

print("Flattened array:", flattened)
print("Original matrix after modifying flatten():")
print(matrix)
```

```python
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

## Q4. Cognitive Assistive System Using OLS

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("X:")
print(X)

print("y:")
print(y)
```

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
X_T = X.T

print("X transpose:")
print(X_T)
```

### Meaning of Transpose

The original X matrix has one row for each user and one column for each feature. After transpose, each row represents one feature and the columns represent the users.

```python
XTX = X.T @ X

print("X.T @ X:")
print(XTX)
```

```python
XTX_inv = np.linalg.inv(XTX)

print("Inverse of X.T @ X:")
print(XTX_inv)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("OLS coefficients:")
print(beta)
```

### Interpretation of Coefficients

The three coefficients correspond to sleep hours, activity level, and stress level respectively.

A coefficient represents the estimated change in the assistance score for a one-unit change in that feature while keeping the other features constant.

The sign of a coefficient shows whether the feature has a positive or negative relationship with the predicted assistance score.

```python
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
