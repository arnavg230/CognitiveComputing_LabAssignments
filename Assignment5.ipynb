## Q1. Basic Operations on NumPy Array

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

## Q2(a). Reverse the NumPy Array

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

## Q2(b). Most Frequent Value and Its Indices

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()

    most_frequent_values = values[counts == max_count]

    for value in most_frequent_values:
        indices = np.where(arr == value)[0]
        print("Value:", value)
        print("Frequency:", max_count)
        print("Indices:", indices)
        print()

print("For x:")
most_frequent(x)

print("For y:")
most_frequent(y)
```

## Q3. Accessing Elements of a 2-D Array

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr)

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

## Q4. NumPy Array Using linspace()

```python
ucs420_Arnav = np.linspace(10, 100, 25)

print("Array:")
print(ucs420_Arnav)

print("Dimensions:", ucs420_Arnav.ndim)
print("Shape:", ucs420_Arnav.shape)
print("Total elements:", ucs420_Arnav.size)
print("Data type:", ucs420_Arnav.dtype)
print("Total bytes:", ucs420_Arnav.nbytes)
```

```python
reshaped_Arnav = ucs420_Arnav.reshape(25, 1)

print("Reshaped array:")
print(reshaped_Arnav)

print("Transpose:")
print(reshaped_Arnav.T)
```

### Observation

For a 1-D array, the T attribute does not change the array into a row or column. 
After reshaping the array to 25 x 1, the T attribute can be used to get its transpose.

## Q5. Operations on a 2-D NumPy Array

```python
ucs420_Arnav = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Arnav)
```

```python
print("Mean:", np.mean(ucs420_Arnav))
print("Median:", np.median(ucs420_Arnav))
print("Maximum:", np.max(ucs420_Arnav))
print("Minimum:", np.min(ucs420_Arnav))
print("Unique elements:", np.unique(ucs420_Arnav))
```

```python
reshaped_ucs420_Arnav = ucs420_Arnav.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_Arnav)
```

```python
resized_ucs420_Arnav = np.resize(ucs420_Arnav, (2, 3))

print("Resized array:")
print(resized_ucs420_Arnav)
```
