```python
roll_no = 1024170057

L = [int(i) * 10 for i in str(roll_no)]

print(L)

L.append(90)
print(L)

L.insert(2, 50)
print(L)

L.remove(90)
print(L)

L.pop()
print(L)

L.sort()
print(L)

L.sort(reverse=True)
print(L)

print(L[:3])
print(L[-3:])

avg = sum(L) / len(L)

new_list = [i for i in L if i > avg]

print(new_list)
```

```python
scores = tuple(L[:8])

print(scores)

highest = max(scores)
lowest = min(scores)

print(highest, scores.index(highest))
print(lowest, scores.count(lowest))

# Tuples are immutable, so they cannot be reversed in place.
reverse_scores = list(scores[::-1])

print(reverse_scores)

score = int(input("Enter score: "))

if score in scores:
    print(scores.index(score))
else:
    print("Score not present")

# scores[0] = 100
# TypeError occurs because tuple elements cannot be modified after creation.

first, second, *remaining = scores

print(first, second, remaining)
```

```python
import random

random.seed(1024170057)

numbers = [random.randint(100, 900) for i in range(100)]

odd_numbers = [i for i in numbers if i % 2 != 0]

even_numbers = [i for i in numbers if i % 2 == 0]

print(odd_numbers)
print(len(odd_numbers))

print(even_numbers)
print(len(even_numbers))


def prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


prime_numbers = [i for i in numbers if prime(i)]

print(prime_numbers)
print(len(prime_numbers))


count = {}

for i in numbers:
    count[i] = count.get(i, 0) + 1

most_frequent = max(count, key=count.get)

print(most_frequent, count[most_frequent])
```

```python
digits = [int(i) for i in str(1024170057)]

A = {i * 7 for i in digits}

B = {i * 9 for i in digits}

print(A)
print(B)

print(A.union(B))

print(A.intersection(B))

print(A.difference(B))

print(B.difference(A))

# difference gives separate values from one set while symmetric difference gives all non-common values.

print(A.symmetric_difference(B))

print(A.issubset(B))

print(B.issuperset(A))

x = int(input("Enter value to remove from A: "))

A.discard(x)

print(A)

# discard does not give an error if the value is not present.
```

```python
my_dict = {
    "name": "Arnav Gupta",
    "roll_no": "1024170057",
    "branch": "COPC",
    "age": 20,
    "city": "Delhi"
}


my_dict["location"] = my_dict.pop("city")

my_dict["cgpa"] = 8.15

my_dict["age"] = my_dict["age"] + 1


dict_pop = my_dict.copy()

dict_del = my_dict.copy()

dict_pop.pop("branch")

del dict_del["branch"]

# pop returns the deleted value while del only removes the key.


for key, value in my_dict.items():
    print(key, "→", value)


if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email not available")


friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024170060",
    "branch": "COPC",
    "age": 20,
    "city": "Mumbai"
}


merged_dict = {**my_dict, **friend_dict}

print(merged_dict)

# Values of the second dictionary overwrite values of the first dictionary when keys are same.


string_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}

print(string_dict)
```
