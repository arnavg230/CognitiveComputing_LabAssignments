### Q6(e). Statistics

```python
print("Average salary:", employees["Salary"].mean())
print("Total bonus:", employees["Bonus"].sum())
print("Youngest employee age:", employees["Age"].min())
print("Highest performance rating:", employees["Rating"].max())
```

### Q6(f). Sort by Salary in descending order

```python
employees.sort_values(by="Salary", ascending=False)
```

### Q6(g). Performance Category

```python
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(performance)

employees
```

### Q6(h). Missing Values

```python
employees.isnull().sum()
```

### Q6(i). Rename Employee_ID to ID

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)

employees
```

### Q6(k). Add Tax column

```python
employees["Tax"] = employees["Salary"] * 0.10

employees
```
